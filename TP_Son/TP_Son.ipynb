```python
############ 3. Gestion du volume et de la dynamique #############

###### 3.1 Creation d'un crescendo


#1) On augmente de maniere lineaire l'amplitude du son produit de la meme maière que l'on vient de faire pour la fréquence

#2) 
def crescendo_sine(freq,duration,increase):
    temps=np.linspace(0,duration, duration*44100)
    if increase==True:
        amplitudes=np.linspace(0,1,duration*44100)
        son=amplitudes*np.sin(2*np.pi*temps*freq)
    else:
        amplitudes=np.linspace(1,0,duration*44100)
        son=amplitudes*np.sin(2*np.pi*temps*freq)
    plt.plot(temps,son)
    plt.show()

crescendo_sine(440,4,False)
```
